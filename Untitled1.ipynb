```python
import requests

API_KEY = 'sk-6549f06fb6b941cea7442e5451561a58'
API_URL = 'https://api.deepseek.com/v1/chat/completions'

import requests
import ipywidgets as widgets
from IPython.display import display, HTML
from IPython.display import clear_output

# ==========================================================
# CONFIGURACIÓN
# ==========================================================

# PEGA AQUÍ TU NUEVA API KEY
API_KEY = "PEGA_TU_NUEVA_API_KEY_AQUI"

URL = "https://api.openai.com/v1/responses"

# Modelo
MODEL = "gpt-5.6-luna"

# ==========================================================
# ESTILO DE LA PÁGINA
# ==========================================================

display(HTML("""
<style>

body {
    background-color: #050505 !important;
}

.netflix-container {
    background: #080808;
    color: white;
    padding: 25px;
    border-radius: 15px;
    font-family: Arial, sans-serif;
    max-width: 1000px;
    margin: auto;
}

.netflix-title {
    color: #e50914;
    font-size: 40px;
    font-weight: bold;
    margin-bottom: 5px;
}

.netflix-subtitle {
    color: #cccccc;
    font-size: 17px;
    margin-bottom: 25px;
}

.chat-box {
    background: #141414;
    border-radius: 12px;
    padding: 20px;
    margin-top: 15px;
    min-height: 250px;
    max-height: 450px;
    overflow-y: auto;
}

.user-message {
    background: #333333;
    padding: 12px;
    border-radius: 10px;
    margin: 10px 0;
    text-align: right;
}

.bot-message {
    background: #e50914;
    padding: 12px;
    border-radius: 10px;
    margin: 10px 0;
    text-align: left;
}

.section-title {
    color: white;
    font-size: 22px;
    margin-top: 30px;
    margin-bottom: 10px;
}

</style>

<div class="netflix-container">

<div class="netflix-title">
NETFLIX AI
</div>

<div class="netflix-subtitle">
Tu asistente inteligente de películas, series y mucho más 🎬🤖
</div>

<div class="section-title">
🔥 Recomendaciones
</div>

<p>
Pregúntame por películas, series, actores, géneros,
ideas para ver o incluso temas completamente diferentes.
</p>

</div>
"""))

# ==========================================================
# HISTORIAL DE CONVERSACIÓN
# ==========================================================

historial = []

# ==========================================================
# FUNCIÓN PARA HABLAR CON LA IA
# ==========================================================

def preguntar_ia(pregunta):

    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }

    # Convertimos el historial en texto
    contexto = ""

    for mensaje in historial:
        contexto += mensaje + "\n"

    instrucciones = """
Eres Netflix AI, un asistente inteligente y amigable.

Tu función principal es ayudar al usuario con:

- Películas
- Series
- Recomendaciones
- Géneros
- Actores
- Directores
- Historias
- Personajes
- Cultura
- Tecnología
- Programación
- Tareas escolares
- Matemáticas
- Ciencias
- Inglés
- Español
- Ideas
- Preguntas generales

Debes responder de forma clara, entretenida y fácil de entender.

Si el usuario pregunta por una película o serie,
explica brevemente de qué trata y por qué podría interesarle.

No inventes datos cuando no estés seguro.

Si el usuario cambia de tema, síguelo normalmente.

Habla en español salvo que el usuario pida otro idioma.

No digas que solamente puedes hablar de Netflix.
Eres un asistente multiusos.
"""

    entrada = f"""
{instrucciones}

Historial de conversación:
{contexto}

Nueva pregunta del usuario:
{pregunta}
"""

    data = {
        "model": MODEL,
        "input": entrada
    }

    try:

        respuesta = requests.post(
            URL,
            headers=headers,
            json=data,
            timeout=60
        )

        if respuesta.status_code != 200:

            return (
                "❌ Ocurrió un problema con la API.\n\n"
                f"Código de error: {respuesta.status_code}\n"
                f"Detalles: {respuesta.text}"
            )

        resultado = respuesta.json()

        # Obtener el texto de la respuesta
        texto = resultado.get("output_text")

        if texto:
            return texto

        # Método alternativo por si output_text no aparece
        texto_final = ""

        for item in resultado.get("output", []):

            for contenido in item.get("content", []):

                if contenido.get("type") == "output_text":
                    texto_final += contenido.get("text", "")

        if texto_final:
            return texto_final

        return "🤖 No recibí una respuesta de la IA."

    except requests.exceptions.Timeout:

        return "⏳ La solicitud tardó demasiado. Intenta nuevamente."

    except requests.exceptions.ConnectionError:

        return "🌐 No se pudo conectar con Internet."

    except Exception as e:

        return f"❌ Error: {str(e)}"


# ==========================================================
# INTERFAZ DEL CHAT
# ==========================================================

chat = widgets.Output()

entrada = widgets.Text(
    placeholder="Escribe aquí tu pregunta...",
    layout=widgets.Layout(width="80%", height="45px")
)

boton = widgets.Button(
    description="ENVIAR",
    button_style="danger",
    layout=widgets.Layout(width="18%", height="45px")
)

boton_limpiar = widgets.Button(
    description="🗑️ Limpiar",
    layout=widgets.Layout(width="100px")
)


# ==========================================================
# MOSTRAR MENSAJES
# ==========================================================

def mostrar_mensaje(usuario, respuesta):

    with chat:

        display(HTML(f"""
        <div style="
            background:#333;
            color:white;
            padding:12px;
            margin:8px;
            border-radius:10px;
            text-align:right;">
            <b>👤 Tú:</b><br>
            {usuario}
        </div>
        """))

        display(HTML(f"""
        <div style="
            background:#e50914;
            color:white;
            padding:12px;
            margin:8px;
            border-radius:10px;
            text-align:left;">
            <b>🤖 Netflix AI:</b><br>
            {respuesta.replace(chr(10), '<br>')}
        </div>
        """))


# ==========================================================
# BOTÓN ENVIAR
# ==========================================================

def enviar_mensaje(b):

    pregunta = entrada.value.strip()

    if pregunta == "":
        return

    entrada.value = ""

    respuesta = preguntar_ia(pregunta)

    # Guardar conversación
    historial.append(f"Usuario: {pregunta}")
    historial.append(f"Netflix AI: {respuesta}")

    mostrar_mensaje(pregunta, respuesta)


# ==========================================================
# LIMPIAR CHAT
# ==========================================================

def limpiar_chat(b):

    historial.clear()

    with chat:
        clear_output()

    print("🧹 Conversación eliminada.")


boton.on_click(enviar_mensaje)

boton_limpiar.on_click(limpiar_chat)


# ==========================================================
# MOSTRAR INTERFAZ
# ==========================================================

display(
    HTML("""
    <div style="
        background:#141414;
        color:white;
        padding:15px;
        border-radius:10px;
        margin-top:20px;
        font-family:Arial;">

        <b>💡 Ejemplos:</b><br><br>

        🎬 Recomiéndame una película de acción<br>
        📺 ¿Qué series de ciencia ficción me recomiendas?<br>
        🧠 Explícame qué es la inteligencia artificial<br>
        📚 Ayúdame con una tarea de matemáticas<br>
        💻 Explícame Python como principiante<br>
        🌎 ¿Cuál es la capital de Colombia?

    </div>
    """)
)

display(chat)

display(
    widgets.HBox(
        [entrada, boton],
        layout=widgets.Layout(width="100%")
    )
)

display(boton_limpiar)
```
